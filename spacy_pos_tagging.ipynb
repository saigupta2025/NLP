```python
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')  # load a specific, pre-trained English language processing pipeline into your Python environment  # en: english, core: general-purpose model , web: corpus of written web text , sm: refers the size as small ;
```

```python
doc = nlp("sai will google about twitter")
print(len(doc))
```

```python
doc.text
```

```python
doc[-1]
```

```python
doc[2].pos_   # _ indicates hash
```

```python
doc[2].tag_
```

```python
spacy.explain('VB')
```

```python
# for word in doc:
#     print(word.text,"------>", word.pos_,word.tag_,spacy.explain(word.tag_))

print(f"{'Token':{15}} {'POS_ Tag':{12}} {'Detailed Tag':{12}} {' Tag Explaination':{15}} ")
print("-" * 80)
for token in doc:
    print(f"{token.text:{15}} {token.pos_:{12}} {token.tag_:{12}} {spacy.explain(token.tag_):{15}}")
```
